# Final Data Merging: Sentiment and Returns
This script merges stock return data with sentiment results for the test set based on ticker and date -> Data preparation for the final result evaluation!

```python
import numpy as np
import pandas as pd
```

```python
df_sentiment = pd.read_csv("sentiments_group 6.csv")
df_sentiment
```

```python
df_sentiment['date_day'] = pd.to_datetime(df_sentiment['date_day'])
df_sentiment = df_sentiment.sort_values(by='date_day')
df_sentiment
```

```python
df_sentiment["sentiment"].value_counts()
```

```python
df_return = pd.read_csv("train_return_data_daily.csv")
df_return
```

```python
# sentiment
df_sentiment['date_day'] = pd.to_datetime(df_sentiment['date_day'])
df_sentiment['Ticker'] = df_sentiment['Ticker'].str.upper()

# return
df_return['date'] = pd.to_datetime(df_return['date'])
df_return['TICKER'] = df_return['TICKER'].str.upper()
```

```python
df_sent_daily = (
    df_sentiment
    .groupby(['Ticker', 'date_day'], as_index=False)
    .agg(sentiment_daily=('sentiment', 'mean'))
)
```

```python
df_merged = (
    df_return
    .merge(
        df_sent_daily,
        left_on=['TICKER', 'date'],
        right_on=['Ticker', 'date_day'],
        how='right'
    )
)
```

```python
df_merged['sentiment_daily'] = df_merged['sentiment_daily'].fillna(0)
df_merged
```

```python
df_merged["sentiment_daily"].value_counts()
```

```python
df_merged.to_csv("daily_sentiment_return.csv", index=False)
```
